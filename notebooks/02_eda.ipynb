# Notebook 02: Exploratory Data Analysis & Failure Window Analysis
Understanding sensor behavior before machine failure and identifying feature engineering opportunities.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/raw/machine_predictive_maintenance.csv')
df['timestamp'] = pd.to_datetime(df['timestamp'])

# Machine failure class distribution
print('Failure label balance:\n', df['failure_within_24h'].value_counts(normalize=True))
print('\nFailure types:\n', df['failure_type'].value_counts())
```

```python
# Sensor correlation with failure label
numeric_cols = ['temperature', 'vibration', 'pressure', 'current', 'rpm', 'load', 'operating_hours', 'failure_within_24h']
corr = df[numeric_cols].corr()
print('Correlation with failure_within_24h:\n', corr['failure_within_24h'].sort_values(ascending=False))
```

```python
# Sensor trajectories around Machine M17 failure window
m17 = df[df['machine_id'] == 'M17'].sort_values('timestamp').reset_index(drop=True)
plt.figure(figsize=(12, 4))
plt.plot(m17['vibration'], label='Vibration (mm/s)', color='crimson')
plt.title('M17 Vibration Escalation Leading to Bearing Failure')
plt.legend()
plt.show()
```
